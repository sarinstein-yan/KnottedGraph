# Non-Hermitian Nodal Skeletons

This notebook shows the canonical application API. The root `knotted_graph` package is generic; the nodal-skeleton workflow lives under `knotted_graph.applications.nodal`.

```python
import sympy as sp

from knotted_graph.applications.nodal import NodalSkeleton
from knotted_graph.applications.nodal.models import hopf_link_bloch_vector

kx, ky, kz = sp.symbols("k_x k_y k_z", real=True)
bloch_vector = hopf_link_bloch_vector(0.1, k_symbols=(kx, ky, kz))
```

For interactive exploration, use a larger grid. For a quick smoke run, keep the dimension small.

```python
ske = NodalSkeleton(
    bloch_vector,
    k_symbols=(kx, ky, kz),
    dimension=64,
)

ske.is_Hermitian, ske.is_PT_symmetric
```

The extracted skeleton graph follows the generic spatial-graph contract: node `pos` attributes and edge `pts` polylines.
