# KnottedGraph User Guide

`KnottedGraph` is organized around one main idea:

$$
\text{3D object}
\rightarrow
\text{embedded spatial graph}
\rightarrow
\text{planar projection}
\rightarrow
\text{PD code}
\rightarrow
\text{Yamada polynomial}.
$$

Choose the notebook that matches where your data enters this pipeline.

| Notebook | Use it when... |
| --- | --- |
| **[01 — Getting Started](01_getting_started.ipynb)** | you want the shortest complete example |
| **[02 — Core Workflows](02_core_workflows.ipynb)** | you need to control graph extraction, minimum-edge contraction, smoothing, projection selection, or PD-code construction |
| **[Physics Applications](applications/01_physics_applications.ipynb)** | your starting point is a physical model, exceptional/nodal surface, vector field, parameter sweep, or material surface |
| **[Mathematics Applications](applications/02_mathematics_applications.ipynb)** | your starting point is a graph family and you want exact invariants, scans, or datasets |
| **[Protein Applications](applications/03_protein_applications.ipynb)** | you are working on the protein workflow; this page is currently a Kehan-owned template |
| **[03 — Advanced & Reproduction](03_advanced_and_reproduction.ipynb)** | you need extraction robustness checks, projection diagnostics, Yamada state-level inspection, or geometric reproducibility tests |
| **[01 — Sanity Checks](benchmarks/01_sanity_checks.ipynb)** | you want to reproduce published Yamada results and compare the supported evaluation routes |
| **[02 — Application Output Regression](benchmarks/02_application_output_regression.ipynb)** | you want to verify that application-level graph, PD-code, and Yamada outputs remain unchanged |
| **[03 — KnottedGraph vs Topoly Scaling](benchmarks/03_knottedgraph_vs_topoly_scaling.ipynb)** | you want the correctness-gated identical-PD Yamada-engine scaling comparison |

If this is your first use of the library, start with **01 — Getting Started**.

## Choose the workflow by your starting object

### I have a 3D surface or physical model
Go to **Physics Applications** after you understand the basic pipeline.

### I already have an embedded graph
Go to **Core Workflows** to inspect cleanup, projection, PD code, and Yamada calculation.

### I want to study a mathematical graph family
Go directly to **Mathematics Applications**.

### I have protein data
The protein-specific graph construction is not yet implemented in this guide. Use
**Protein Applications** as the required interface/template and complete the
domain-specific mapping before running topology.

### I am checking correctness or numerical reliability
Use **01 — Sanity Checks** to compare Yamada calculations directly with published
results. Use **02 — Application Output Regression** to verify application-level
outputs against the reference branch, and **Advanced & Reproduction** for extraction,
projection, and geometric robustness diagnostics.

## The common workflow

Most examples in the library eventually reduce to the same sequence:

$$
\text{geometry or scientific data}
\longrightarrow
G\subset\mathbb{R}^3
\longrightarrow
D(G)
\longrightarrow
\operatorname{PD}(G)
\longrightarrow
\Upsilon(G;Y).
$$

The first step is application-dependent. Once you have an embedded spatial graph
$G$, the projection and invariant tools are shared across applications.

A useful rule when working with the library is to **inspect the graph and the
selected diagram before interpreting the polynomial**. This makes it much easier
to distinguish a genuine topological difference from an extraction or projection
problem.

## How to use these notebooks

Each workflow is written around the same questions:

1. **What should you pass to the library?**
2. **What object does the call return?**
3. **What should you inspect before continuing?**
4. **What topological conclusion can you safely draw from the result?**

Run only the sections relevant to your problem. Optional application stacks are
called out where they are needed, and any section that is not yet implemented is
explicitly marked rather than presented as working API.
