# KnottedGraph User Guide

`KnottedGraph` is organized around one main idea:

$$
\text{3D object}
\rightarrow
\text{embedded spatial graph}
\rightarrow
\text{planar projection}
\rightarrow
\text{PD code}
\rightarrow
\text{Yamada polynomial}.
$$

## Choose the workflow by your starting object

### I have a 3D surface or physical model
Go to **Physics Applications** after you understand the basic pipeline.

### I have a knot/link type, braid, or analytic complex field
Go to **Analytic Knot Fields** to construct exact/reference or braid-derived fields, validate tubular handlebodies, convert them to spatial graphs, and scan knot-field deformations. Physical Bloch-vector deformations are documented there as a separate periodic-model layer.

### I already have an embedded graph
Go to **Core Workflows** to inspect cleanup, projection, PD code, and Yamada calculation.

### I want to study a mathematical graph family
Go directly to **Mathematics Applications**.

### I have protein data
Go to **Protein Applications** for PDB/mmCIF crosslink parsing, crosslink-supported cyclic-core extraction, exact fingerprint perturbations, null rewiring, and optional topology-validated Repulsor preprocessing. Use **Protein Crosslink Topology** for dataset-scale runs.

### I am checking correctness or numerical reliability
Use **Yamada Sanity Checks** for core invariant checks, **Application Regression Checks** for application-level output equality, and **Advanced & Reproduction** for extraction, projection, and geometric robustness diagnostics.


## The common workflow

Most examples in the library eventually reduce to the same sequence:

$$
\text{geometry or scientific data}
\longrightarrow
G\subset\mathbb{R}^3
\longrightarrow
D(G)
\longrightarrow
\operatorname{PD}(G)
\longrightarrow
\Upsilon(G;Y).
$$

The first step is application-dependent. Once you have an embedded spatial graph
$G$, the projection and invariant tools are shared across applications.

A useful rule when working with the library is to **inspect the graph and the
selected diagram before interpreting the polynomial**. This makes it much easier
to distinguish a genuine topological difference from an extraction or projection
problem.

## How to use these notebooks

Each workflow is written around the same questions:

1. **What should you pass to the library?**
2. **What object does the call return?**
3. **What should you inspect before continuing?**
4. **What topological conclusion can you safely draw from the result?**

Run only the sections relevant to your problem. Optional application stacks are
called out where they are needed, and any section that is not yet implemented is
explicitly marked rather than presented as working API.
